```python
Q1 = d['TotalAmount'].quantile(0.25)
Q3 = d['TotalAmount'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = d[(d['TotalAmount'] < lower_bound) | (d['TotalAmount'] > upper_bound)]

print("Number of outliers:", len(outliers_iqr))
display(outliers_iqr)
```

# **Data Preprocessing**
- Delete orders with NaN values
- Rename City.1, Country.1, Phon.1 Columns, to be CompanyCity, CompanyCountry, CompanyPhon
- Change OrderDate Type into Date
- Change ID Columns, and OrderNumber column to integer
- Drop dupliacted rows based on OrderID, and OrderNumber
- Strip and lowercase columns names
- select only these columns: 'customerid', 'firstname', 'lastname', 'city', 'country', 'orderid', 'ordernumber', 'orderdate', 'totalamount'

```python
d.dropna(subset=['OrderID', 'OrderNumber'], inplace=True)
d.reset_index(drop=True, inplace=True)
```

```python
d.isnull().sum()
```

```python
d.info()
```

```python
d.rename(columns={
    'City.1': 'CompanyCity',
    'Country.1': 'CompanyCountry',
    'Phon.1': 'CompanyPhone'
}, inplace=True)
```

```python
d['OrderDate'] = d['OrderDate'].astype('datetime64[ns]')
```

```python
d['CustomerID'] = d['CustomerID'].astype('int')
d['ProductID'] = d['ProductID'].astype('int')
d['OrderID'] = d['OrderID'].astype('int')
d['SupplierID'] = d['SupplierID'].astype('int')
d['OrderNumber'] = d['OrderNumber'].astype('int')
```

```python
d.drop_duplicates(subset=['OrderID', 'OrderNumber'], inplace=True)
```

```python
d.columns = d.columns.str.strip().str.lower()
```

```python
d_new = d.loc[: , ['customerid', 'firstname', 'lastname', 'city', 'country', 'orderid', 'ordernumber', 'orderdate', 'totalamount']]
```

# **Analysis**


## Univariate Analysis : make analysis for each column

```python
plt.hist(d_new['totalamount'],color='skyblue', edgecolor='black', label = 'Total Amount')
plt.title('Total Amount Distribution')
plt.axvline(d_new['totalamount'].mean(), color='red', linestyle='dashed', label = 'Mean')
plt.axvline(d_new['totalamount'].median(), color='green', linestyle='dashed', label = 'Median')
plt.legend()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=d_new, x='city', order=d_new['city'].value_counts().index)

plt.title('Number of Transactions by City')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.pie(d_new['country'].value_counts(), labels=d_new['country'].value_counts().index, autopct='%1.1f%%')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.bar(d_new['country'].value_counts().index, d_new['country'].value_counts())
plt.xticks(rotation=90)
plt.show()
```

## Bivariate Analysis : make analysis for each two column
- Total Amount per Country using Bar Chart
- Total Amount per City using Bar Chart
- Total Amount per Orderdate using line chart
- Total Amount per Customer full name using bar chart (top ten customers)

```python
total_amount_per_country = d_new.groupby('country')['totalamount'].sum().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
plt.bar(total_amount_per_country.index, total_amount_per_country.values)
plt.title('Total Amount per Country')
plt.xticks(rotation=90)
plt.show()
```

```python
total_amount_per_city = d_new.groupby('city')['totalamount'].sum().sort_values(ascending=False)

plt.figure(figsize=(15, 9))
plt.bar(total_amount_per_city.index, total_amount_per_city.values)
plt.title('Total Amount per city')
plt.xticks(rotation=90)
plt.show()
```

```python
total_amount_per_date = d_new.groupby('orderdate')['totalamount'].sum()
plt.figure(figsize = (10, 5))
plt.plot(total_amount_per_date.index, total_amount_per_date.values)
plt.title('Total Amount per Date')
plt.show()
```

```python
total_amount_per_full_name = d_new.groupby(['firstname', 'lastname'])['totalamount'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(12, 6))
plt.bar(total_amount_per_full_name.index.map(lambda x: f'{x[0]} {x[1]}'), total_amount_per_full_name.values)
plt.title('Top 10 Customers by Total Amount')
plt.xticks(rotation=90)
plt.show()
```
